```python
#8.	Write a Python program using a while loop to calculate the factorial of a number.
num = int(input("Enter a number: "))
if num<0:
    print("Factorial is not defined for negative numbers.")
else:
    factorial=1
    counter=num
    while counter>1:
        factorial*=counter
        counter-=1
    print(f"The factorial of {num} is {factorial}")
```

```python
#10.	Write a Python program using a while loop to find the sum of the digits of a given number.
num=int(input("Enter a number: "))
n=abs(num)
digit_sum=0
while n>0:
    digit=n%10
    digit_sum+=digit
    n=n//10
print(f"The sum of the digits of {num} is: {digit_sum}")
```

```python
#11. Write a Python program using a while loop to reverse a given number.
number=int(input("Enter number to be reversed: "))
reversed_num=0
while number>0:
    digit=number%10
    reversed_num=(reversed_num * 10)+digit
    number=number//10
print("Reversed number:",reversed_num)
```

```python
#14.	Write a Python program using a while loop to check whether a given number is a palindrome.
number = int(input("Enter number: "))
original_number = number
reversed_num = 0
while number > 0:
    digit = number % 10
    reversed_num = (reversed_num * 10) + digit
    number = number // 10
if original_number == reversed_num:
    print(f"{original_number} is a palindrome.")
else:
    print(f"{original_number} is not a palindrome.")
```

```python
#15.	Write a Python program using a for loop to calculate the sum of all odd numbers between two numbers entered by the user.
start = int(input("Enter the start number: "))
end = int(input("Enter the end number: "))
total_sum = 0
for num in range(start, end + 1):
    if num % 2 != 0:
        total_sum += num
print("Sum of odd numbers:", total_sum)
```

```python
#16.	Write a Python program using a while loop to repeatedly take a number from the user until the user enters 0, then display the total sum.
total_sum = 0
number = int(input("Enter a number (0 to stop): "))
while number != 0:
    total_sum += number
    number = int(input("Enter a number (0 to stop): "))
print("Total sum:", total_sum)
```

```python
#18.	Write a Python program using a while loop to find the largest digit in a given number.
number = int(input("Enter number: "))
largest_digit = 0
while number > 0:
    digit = number % 10
    if digit > largest_digit:
        largest_digit = digit
    number = number // 10
print("Largest digit:", largest_digit)
```
